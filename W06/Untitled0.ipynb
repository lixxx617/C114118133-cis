```python
# 匯入資料科學的核心套件
import numpy as np
import pandas as pd

# 設定亂數種子以確保結果可重現
random_seed = 42
rng = np.random.default_rng(random_seed)

print("NumPy 與 Pandas 套件匯入成功！")
print("亂數產生器已準備就緒，種子值為：", random_seed)
```

```python
# 設定學生總數
student_count = 60
n = student_count

# 1. 學生 ID
student_id = np.arange(1, n + 1)

# 2. 科系（隨機抽樣）
departments_list = ["資訊管理", "企業管理", "金融管理"]
department = rng.choice(departments_list, size=n)

# 3. 睡眠時數
sleep_hours = np.round(rng.normal(loc=6.7, scale=1.1, size=n), 1)

# 4. 螢幕使用時數（使用 clip 限制上下限）
screen_time_hours = rng.normal(loc=5.0, scale=1.5, size=n)
screen_time_hours = np.round(np.clip(screen_time_hours, 1.0, 10.0), 1)

# 5. 讀書時數
study_hours = rng.normal(loc=2.5, scale=1.0, size=n)
study_hours = np.round(np.clip(study_hours, 0.5, 6.0), 1)

# 6. 出席率
attendance_rate = rng.normal(loc=88, scale=8, size=n)
attendance_rate = np.round(np.clip(attendance_rate, 60, 100)).astype(int)

# 組合成長方形的 DataFrame 表格
df = pd.DataFrame({
    "student_id": student_id,
    "department": department,
    "sleep_hours": sleep_hours,
    "screen_time_hours": screen_time_hours,
    "study_hours": study_hours,
    "attendance_rate": attendance_rate
})

# 顯示前五筆資料
df.head()
```

```python
# 產生平均為 0、標準差為 5 的隨機誤差
noise = rng.normal(loc=0, scale=5, size=n)

# 逐步計算小考成績
calculated_quiz = (
    50
    + 3 * df["study_hours"]
    + 2 * df["sleep_hours"]
    - 1.2 * df["screen_time_hours"]
    + 0.15 * df["attendance_rate"]
    + noise
)

# 限制分數在 0~100 並四捨五入
df["quiz_score"] = np.round(np.clip(calculated_quiz, 0, 100)).astype(int)

# 選取部分欄位並顯示前五筆
selected_columns = [
    "student_id",
    "study_hours",
    "sleep_hours",
    "screen_time_hours",
    "attendance_rate",
    "quiz_score"
]

df[selected_columns].head()
```
